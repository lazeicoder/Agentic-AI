```python
from pydantic import BaseModel
```

```python
class Agent(BaseModel):
    name: str
    age: int
    city: str 

agent = Agent(name="OpenAI", age=15, city="California")
print(agent)
```

```python
from typing import Optional

class Person(BaseModel):
    name: str 
    age: int 
    city: str 
    salary: Optional[float] = None # optional with default value None
    is_active: Optional[bool] = True # optinal with default value True

person = Person(name="Anurag", age=23, city="San Francisco")
print(person)
```

```python
from typing import List 

class Person(BaseModel):
    room_number: int 
    students: List[str] # list of strings
    roll_number: List[int] # list of integer numbers
    capacity: int 

person = Person(room_number=3, students=('Anurag', 'Mayukh', 'Rupam', 'Srini'), roll_number=(4, 17, 78, 43), capacity=4)
print(person)
```

```python
class ClassRoom(BaseModel):
    rm: int
    studs: List[str]
    roll: List[int]
    cap: int 

classroom = ClassRoom(rm=5, studs=["Anurag", "Charlie", "Bob", "Shivali"], roll=[3, 23, 16, 19], cap=8)
# print(classroom)

try:
    classroom = ClassRoom(rm=5, studs=["Anurag", "Charlie", "Bob", 89], roll=[3, 23, 16, 19], cap=8)
except ValueError as e:
    print(f"Error: {e}")
```

```python
# Nested Base Models

class Address(BaseModel):
    street: str
    city: str 
    zip_code: int

class Customer(BaseModel):
    customer_id: int 
    name: str 
    address: Address # nested model

customer = Customer(
    customer_id=1,
    name="Harkirat Singh",
    address=Address(
        street="Gold Street",
        city="Bangalore",
        zip_code=700265
    )
)

print(customer)
```

```python
# fields in pydantic: extra layer of data validation
from pydantic import Field

class PassAllocation(BaseModel):
    name: str = Field(min_length=8, max_length=25)
    avg_marks: float = Field(gt=36, le=92)
    quality: int = Field(ge=0)


p1 = PassAllocation(
    name= "Hasnan Ahmed",
    avg_marks= 46.98,
    quality= 74
)

print(p1)
```

```python
class User (BaseModel):
    username: str = Field(..., description="Unique username for user")
    age: int = Field(default=18, description="User age, defaults to 18")
    email: str = Field(default_factory=lambda: "username@gmail.com", description="Default email address")


user1 = User(
    username="Alisha"
)

print(user1)

user2 = User(
    username="Harshadd",
    age=43,
    email="harshadd@prism.ai"
)

print(user2)
```
