```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

loader = TextLoader("speech.txt")
documents = loader.load()


text_splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap=20)
docs = text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = (
    OllamaEmbeddings(model="nomic-embed-text:latest")
)
embeddings
```

```python
db = FAISS.from_documents(docs, embeddings)
db
```

```python
query = "Why does the speaker believe that small actions are powerful?"

docs = db.similarity_search(query)
docs[3].page_content
```

```python
retriever = db.as_retriever()
pages = retriever.invoke(query)

for page in pages:
    print(page.page_content, end="\n-------------------\n")
```

```python
docs_with_score = db.similarity_search_with_score(query)

docs_with_score
```

```python
embedding_vector = embeddings.embed_query(query)

embedding_vector
```

```python
docs_with_vector = db.similarity_search_by_vector(embedding_vector)

docs_with_vector
```

```python
# Building a Sample vectorDB

from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings

# Text Loading from Document
loader = TextLoader("speech.txt")
documents = loader.load()

# Document Splitting into smaller chunks of texts
text_splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap=0)
docs = text_splitter.split_documents(documents)

docs
```

```python
# Text chunks being Embedded

embeddings = (
    OllamaEmbeddings(model="nomic-embed-text:latest")
)

embeddings
```

```python
chromaDB = Chroma.from_documents(docs, embeddings)
chromaDB
```

```python
# Query it

query = "Why does the speaker believe that small actions are powerful?"

results = chromaDB.similarity_search(query)

i = 1
for result in results:
    print(f"Intended results are R:{i} : {result.page_content}\n")
    i = i+1
```

```python
# Saving to the disk

vectordb = Chroma.from_documents(docs, embeddings, persist_directory="./chroma_db")
```

```python
# Loading from the disk

db2 = Chroma(persist_directory="./chroma_db", embedding_function=embeddings)

res = db2.similarity_search(query)
print(res[1].page_content)
```

```python
# Retriever

retriever = db2.as_retriever()
retriever.invoke(query)[1].page_content
```
