```python
# TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")
docs = loader.load()
```

```python
speech = ""
with open("speech.txt") as f:
    speech = f.read()

# print(speech) 

text_split = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=10)
text = text_split.create_documents([speech])

# print(len(text))

for i in range(len(text)):
    print(text[i].page_content)
```

```python
# Character Text Splitter 

from langchain_text_splitters import CharacterTextSplitter

text_split = CharacterTextSplitter(separator="\n\n", chunk_size=100, chunk_overlap=10)
texts = text_split.split_documents(docs)
texts
```

```python
import json
import requests

json_data = requests.get("https://api.smith.langchain.com/openapi.json").json()
print(json_data)
```

```python
from langchain_text_splitters import RecursiveJsonSplitter

json_splitter = RecursiveJsonSplitter(max_chunk_size=200)
json_chunks = json_splitter.split_json(json_data)

print(json_chunks)
```

```python
for chunk in json_chunks[:3]:
    print(chunk)
```

```python
# Json splitter can also output documents

docs = json_splitter.create_documents(texts=[json_data])
for doc in docs[:3]:
    print(doc)
```
