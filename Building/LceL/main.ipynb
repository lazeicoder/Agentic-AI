```python
import os
from dotenv import load_dotenv

load_dotenv()

groq_api_key = os.getenv("GROK_API_KEY")
groq_api_key
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="llama-3.1-8b-instant", groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

# List of messages -> provided to the LLM Model
question = input("Enter your question: ")
messages = [
    SystemMessage(content="You are an expert AI Engineer with over 30+ years of experience in the Software and AI Research domain, with large scale understanding of how AI (different machine learning, deep learning, reinforcement learning techniques, transformers, agentic ai) systems work, perform, architecture, and how to scale them for larger client and user base. So based on your experience in the field, provide me with answers."),
    HumanMessage(content=question)
]

response = model.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
result = output_parser.invoke(response)

print(result)
```

```python
# Chain the components

chain = model | output_parser 
chain.invoke(messages)
```

```python
# Prompt Templates

from langchain_core.prompts import ChatPromptTemplate 

total_prompt = "You are an expert AI Engineer with over 30+ years of experience in the Software and AI Research domain, with large scale understanding of how AI (different machine learning, deep learning, reinforcement learning techniques, transformers, agentic ai) systems work, perform, architecture, and how to scale them for larger client and user base."

generic_template = "Based on your experience in the field, provide me with answers mathematically."

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", generic_template),
        ("user", "{text}"),
    ]
)

```

```python
text = input("Enter text: ")

result = prompt.invoke({
    "text": text
})
```

```python
result.to_messages()
```

```python
# Chaining all components together with LCEL
chain = prompt | model | output_parser
chain.invoke({
    "text": "GenAI"
})
```
